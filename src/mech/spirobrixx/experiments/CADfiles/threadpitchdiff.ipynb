```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
x = np.arange(-30,30, 0.1)
y1 = np.cos(x) 
y2 = np.cos(x*1.02)-0.5
plt.rcParams['figure.figsize'] = [15, 5]
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,y1)
ax.fill_between(x,y1,1.2)
ax.plot(x,y2)
ax.fill_between(x,y2,-1.7)
```

```python
x = np.arange(-30,30, 0.1)
y1a = np.cos(x)
xb = x +0.3
y1b = np.cos(x) 
y2 = np.cos(x*1.02)-0.5
plt.rcParams['figure.figsize'] = [15, 5]
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x[:301],y1a[:301])
ax.fill_between(x[:301],y1a[:301],1.2)
ax.plot(x[:400],y2[:400])
ax.fill_between(x[:400],y2[:400],-1.7)
ax.plot(xb[300:],y1b[300:])
ax.fill_between(xb[300:],y1b[300:],1.2)
```
